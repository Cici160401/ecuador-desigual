```python
import pandas as pd

df = pd.read_csv(
    "../data/raw/archivos/3.1.csv",
    encoding="cp1252",
    header=None
)
```

```python
df_datos = df.iloc[11:].copy()

df_datos.head(10)
```

```python
df_el_oro = df_datos[df_datos[1] == "El Oro"]
df_el_oro.head(15)
```

```python
df_cantones = df_el_oro[
    (df_el_oro[3].str.startswith("Total")) &
    (df_el_oro[2] != "Total El Oro")
]
df_cantones.head(15)
```

```python
print("Número de cantones:", len(df_cantones))

df_cantones[[2, 3, 4, 5]]
```

vamos a calcular el porcentaje de viviendas conectadas a la red pública de alcantarillado.

```python
df_cantones.dtypes
```

```python
df_cantones[4] = pd.to_numeric(
    df_cantones[4].str.replace(",", "", regex=False),
    errors="raise"
)
```

```python
df_cantones[[4, 5]].dtypes
```

```python
df_cantones[5] = pd.to_numeric(
    df_cantones[5].str.replace(",", "", regex=False),
    errors="raise"
)
```

```python
df_cantones[[4, 5]].dtypes
```

```python
# Indicador: Porcentaje de viviendas particulares ocupadas
# conectadas a la red pública de alcantarillado.
# Fuente: INEC, Censo de Población y Vivienda 2022.
# Unidad de análisis: vivienda particular ocupada.
# Desagregación: cantón.

df_cantones["cobertura_pct"] = (df_cantones[5] / df_cantones[4]) * 100

#redondear
df_cantones["cobertura_pct"] = df_cantones["cobertura_pct"].round(2)

#nombre de las columnas
df_cantones = df_cantones.rename(columns={
    2: "canton",
    4: "total_viviendas",
    5: "viviendas_alcantarillado"
})
```

```python
df_cantones[
    ["canton", "total_viviendas",
     "viviendas_alcantarillado", "cobertura_pct"]
]
```

```python
#canton con menor cobertura de alcantarillado
df_cantones[df_cantones["cobertura_pct"] == df_cantones["cobertura_pct"].min()]
```

```python
#cantones ordenados de menor a mayor cobertura de alcantarillado
df_cantones.sort_values(by="cobertura_pct", ascending=True)
```
