```python
import pandas as pd

df_agua = pd.read_csv(
    "../data/raw/archivos/2.1.csv",
    encoding="cp1252",
    header=None
)

df_agua.iloc[:15]
```

```python
pd.set_option("display.max_columns", None)

df_agua.iloc[7:20]
```

```python
df_agua_datos = df_agua.iloc[11:].copy()
```

```python
df_agua_el_oro = df_agua_datos[
    df_agua_datos[1] == "El Oro"
].copy()

df_agua_el_oro[[1, 2, 3, 4, 5, 6, 7]].head(15)
```

```python
condicion_total = (
    df_agua_el_oro[3]
    == "Total " + df_agua_el_oro[2]
)
```

```python
condicion_recepcion = (
    df_agua_el_oro[4]
    == "Total " + df_agua_el_oro[2]
)
```

```python
df_agua_cantones = df_agua_el_oro[
    condicion_total & condicion_recepcion
].copy()

df_agua_cantones[[2, 3, 4, 5, 6, 7]]
```

```python
for col in [5, 6, 7]:
    df_agua_cantones[col] = pd.to_numeric(
        df_agua_cantones[col].str.replace(",", "", regex=False),
        errors="raise"
    )
```

```python
df_agua_cantones["cobertura_municipal_pct"] = (
    df_agua_cantones[6] / df_agua_cantones[5]
) * 100
```

```python
df_agua_cantones = df_agua_cantones.rename(columns={
    2: "canton",
    5: "total_viviendas",
    6: "viviendas_municipio",
    7: "viviendas_juntas_agua"
})

df_agua_cantones[
    ["canton", "total_viviendas",
     "viviendas_municipio", "cobertura_municipal_pct"]
].round(2)
```
